Count the number of words in the file

Print the 3rd to 5th words from the file.

```
all_words = []
with open('poem.txt', 'r') as file:
    for line in file:
        all_words.extend(line.split())

if len(all_words) >= 5:
    print(f"3rd word: {all_words[2]}")
    print(f"4th word: {all_words[3]}")
    print(f"5th word: {all_words[4]}")
else:
    print("not enough words.")
```

```
import statistics
line_lengths = []
with open('poem.txt','r') as file:
    for line in file:
        line_len = len(line.strip())
        line_lengths.append(line_len)
if line_lengths:
    print(f"{sum(line_lengths)}")
    print(f"{statistics.mean(line_lengths):.2f}")
    print(f"{statistics.median(line_lengths):.2f}")
    print(f"{statistics.stdev(line_lengths):.2f}")
else:
    print("No lines found.")
```

Print the last word from the file and change the second last characted in the string to 'g'.

```
last_word = ""
with open('poem.txt', 'r') as file:
    for line in file:
        words = line.split()
        if words:
            last_word = words[-1]
if last_word:
    if len(last_word) >= 2:
        modified_word = last_word[:-2] + 'g' + last_word[-1]
        print(f"original last word: {last_word}")
        print(f"modified last word (second last char changed to 'g'): {modified_word}")
    else:
        print(last_word)
else:
    print("no words found in poem.txt.")
```

The operation completed successfully. The last word was extracted, and its second to last character was successfully changed. This was possible because strings in Python, while immutable, allow for creating new strings based on slices and concatenations of existing ones.

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

Find the average hwy fuel economy across all cars

```
total_hwy_mpg = 0
car_count = 0
for car in mpg:
    total_hwy_mpg += float(car['hwy'])
    car_count += 1
if car_count > 0:
    average_hwy_mpg = total_hwy_mpg / car_count
    print(average_hwy_mpg)
else:
    print("No car data available to calculate average hwy mpg.")
```

Use set to return the unique values for the number of cylinders the cars.

```
cylinders = set(d['cyl'] for d in mpg)
print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
MpgByCyl = []
for c in cylinders:
    summpg = 0
    typecount = 0
    for d in mpg:
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1
    if typecount > 0:
        MpgByCyl.append((c, summpg / typecount))
MpgByCyl.sort(key=lambda x: x[1])
print(MpgByCyl)
```

Return the unique values for the class types in our dataset using set.

```
vehicleclass = set(d['class'] for d in mpg)
print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```
MileageByClass = []
for t in vehicleclass:
    summpg = 0
    classCount = 0
    for d in mpg:
        if d['class'] == t:
            summpg += float(d['hwy'])
            classCount += 1
    if classCount > 0:
        MileageByClass.append((t, summpg / classCount))
MileageByClass.sort(key=lambda x: x[1])
print(MileageByClass)
```
